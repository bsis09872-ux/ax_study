## 2-2 기본 자료형과 형변환

### 기본 용어 정리
#### 0. 주석 (Comment, e notation)
- 인터프리터가 해석 하지 않음
- 표기 방법: # (한 줄 주석), """..."""(여러 줄 주석)
- 역할: 
    - 코드 상에 설명을 추가할때 사용
    - 코드 실행 배제: 인터프리터가 해석을 하지 않으므로 코드 실행을 배제할 수 있음

#### 1. 파이썬의 기본 자료형 (Primitive Data Types) 
- 자료: 연산에 필요한 값(데이터)
- 자료형: 값의  유형 (데이터의 종류)
    - 숫자 (정수형:int, 실수형:float)
    - 문자 (문자; 문장; str)
         - '' 또는 ""로 감싸서 선언(한 줄), """..."""(여러 줄 문장)
    - 논리형 (참(True)과 거짓(False); bool)
    - 값 없음 (None; 값이 없음을 의미함; NoneType)


#### 2. 명시적 형변환 (Explicit Type Conversion)
- 연산은 같은 자료형끼리 가능

- 형변환 함수
    - int(...): 정수로 변환
    - float(...): 실수로 변환
    - bool(...): T/F로 변환
    - str(...): 문자형으로 변환

#### 3. 묵시적 형변환 (Implicit Type Conversion)
- 업캐스팅
    - 파이썬은 정수와 실수를 연산할 때 표현 범위가 더 넓은 실수형 쪽으로 자동 변환하여 데이터 손실을 방지하는데, 이를 업캐스팅(묵시적 형변환)이라고 합니다. 명시적 형변환은 개발자가 int(), float() 등을 직접 호출하는 경우이며, ValueError는 변환 실패 시 발생하는 에러이므로 오답입니다.

#### 4. 형변환 시 주의할 점; 예외와 에러 (ValueError)

#### 5. 거짓으로 인식하는 값
- 0, "", [], None
    - 0과 빈값들

```python
num1 = 100
num2 = 3.123
```

```python
num3 = num1 + num2
print(num3)
type(num3)
```

```python
# num1은 int형, num2는 float형이므로 num1 + num2는 float형으로 변환되어 계산됨 (묵시적 형변환)
num3 = num1 + num2 
print(num3), type(num3)
```

```python
num3 = num1 + int(num2) # 명시적 형변환
print(num3), type(num3)
```

```python
time = "10시"
if time.isdigit():
    print(int(time) + 10)
else:
    print("문자열을 숫자로 변환할 수 없습니다.")
```
