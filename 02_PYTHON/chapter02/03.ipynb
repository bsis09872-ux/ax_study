## 2-3 연산자

#### 1. 항과 연산자
- 항: 연산에 사용되는 값
    - 2개 -> 2항, 3개-> 3항...연산
- 연산자: 연산에 사용되는 기호 (+, -, =, ...)
#### 2. 연산자의 종류
- 대입 연산자 (=)
    - 연산의 우선 순위가  가장 낮다. 가장 마지막에 연산된다.
- 산술 연산자
    - +,-
    - *(곱하기), /(나누기)
    - // (나누기의 몫), % (나누기의 나머지)
    - ** (거듭제곱)

    - 연산 우선순위: **, *, /, //, % >>> +,-
    - 우선순위 강제 적용: 소괄호 사용 "()"
- 비교 연산자
    - 등호, 부등호: >, <, >=, <=, ==, !=
    - 연산의  결과는 무조건 T/F
    - <<판별>>이 필요한 구문 (조건문, 반복문)
    - "==" 동등성, 동일한 가치

        - 동등성 vs. 동일성
            - 동등성: 동일한 가치를 비교 (==)
            - 동일성: 동일한 대상인지 비교 (is: 동일 대상인지 주소 비교)
- 논리 연산자 (and, or, not)
    - and
        - 각 항이 모두 True -> True, 하나라도 False -> False
    - or
        - 모든 항 중에서 하나라도 True -> True (모든 항이 False 일때만 False)
    - not
        - True 조건 -> False 조건으로 반전, False 조건 -> True 조건으로 반전
        - 비교 연산자 보다 우선순위가 낮다
- 복합 대입 연산자
    - 대입 연산자 (=) + 다른 연산자(주로 산술 연산자) 결합
- 부호 반전 연산자 (-)
    - 음수 -> 양수
    - 양수 -> 음수

#### 3. 단락 회로 평가 (Short-Circuit Eval)
- and 연산은 좌측이 False이면, or 연산은 좌측이 True이면 우측 항을 평가하지 않고 즉시 결과를 확정합니다. 
- 이를 단락 회로 평가라고 하며, 불필요한 연산을 줄이고 에러(예: ZeroDivisionError)를 방지하는 효과가 있습니다.
#### 4. 연산자 우선순위
- 대입 연산자가 가장 낮다.
- 논리 연산자
- 비교 연산자
- 산술 연산자
- (...) 소괄호 우선순위가 가장 높다.

```python
# 3
# 첫 항의 결과가 True이면 두 번째 항의 결과를 확인하지 않고 True를 반환

result = True or 10**2
result
```

```python
# 첫 항의 결과가 False이면 두 번째 항의 결과를 확인하고 그 결과를 반환

result = False or 10**2
result
```

```python
# and 연산자는 첫 항의 결과가 False이면 두 번째 항의 결과를 확인하지 않고 False를 반환
result = False and 10**2
result
```

```python
# and 연산자는 첫 항의 결과가 True이면 두 번째 항의 결과를 확인하고 그 결과를 반환
result = True and 10**2
result
```

```python
# or 연산을 이용한 기본값(없을때 대체되는 값)

message = ""
result = message or "기본 메시지"
result
```
