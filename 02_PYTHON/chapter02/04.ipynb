## 2-4 문자열 형식화

1. 1세대: %연산자
    - %s(문자열)
        *문자와 문자열: 문자는 a,b,c 각각을 말함. 문자열은 apple처럼 문자가 연달아 있는 것을 말함.
        - %숫자s -> 최소 숫자 자리수 만큼 공간을 차지, 남는 공간은 공백
    - %d(정수)
        - %대체문자숫자d -> %05s -> 남는 공백은 0으로 대체
    - %f(실수)
        - %, 2f: 소수점 2자리까지 표시
        - %10.2f: 정수부 8자리, 소수점 자리는 2까지 표시

```python
message = "나의 이름은 %10s고, 나이는 %d입니다."
message % ("철수", 20)
```

```python
message = "주문번호: %05s" % 10
message
```

```python
#message = "숫자: %f" % 10.1234567898
#message = "숫자: %.2f" % 10.1234567898
#message = "숫자: %10.2f" % 10.1234567898
message = "숫자: %010.2f" % 10.1234567898
message
```

```python
message = "숫자: %d" % 10.1234567898
message
```

```python
message = "숫자: %s" % 10.123456789
message
```

2. 2세대: str.format() 메서드 방식
- 순서 지정 교체: {0}, {1}, {2} ...
- 단어 지정 교체: {단어} {단어} ...

```python
#message = "내 이름은 {0}이고, 나이는 {1}입니다."
message = "내 이름은 {1}이고, 나이는 {0}입니다."

message.format("철수", 20)
```

```python
message = "내 이름은 {name}이고, 나이는 {age}입니다."
message.format(name="철수", age=20)
```

```python
message = "내 이름은 {name}이고, 나이는 {0}입니다."
message.format(20, name = "철수")
```
