## 3-2: 매핑 및 집합 자료구조

1. 딕셔너리(dictionary) 자료구조
- key - value 쌍 구조
- key : value를 찾기 위한 키워드, 중복 x, 리스트 사용 불가
- value: key로 찾는 값, 중복 o
- {"키": 값}

- 추가(CREATE): 없는 키 이름으로 값을 대입하면 추가됨
- 조회(READ): ['키 이름'] - 키가 없으면 KeyError가 발생
    - get("키 이름") - 키가 없으면 None 반환
    - get("키 이름", "기본값") - 키 이름으로 조회가 되지 않으면 설정된 기본값으로 대체
    - keys(): 모든 키를 담은 veiw, 읽기 전용
    - values(): 모든 값을 담은 view, 읽기 전용
    - items(): (키, 값) 목록을 담은 view
- 수정(UPDATE): 이미 있는 키값에 대입하면 수정
    - update([key: value, key: value, ...])
- 삭제(DELETE):
    - del items['키']
    - clear(): 전체 비우기

- 다른 키-값 쌍 자료형 -> 딕셔너리로 변환
    - dict()

```python
user = {
        "email":"test@test.org",
        "password": 1234, 
        "name":"김철수"
    }
```

```python
user['mobile'] = "010-1000-1000"

user
```

```python
user['name']
```

```python
user.keys()
```

```python
user.values()
```

```python
user.items()
```

```python
user['mobile']
```

```python
mobile = user.get("mobile")
mobile, type(mobile)
```

```python
mobile = user.get("mobile", "없음")
mobile, type(mobile)
```

```python
del user['mobile']
```

### 3. 집합 (set) 자료구조
- 중복을 허용하지 않는다.
- 순서가 결정되지 않는다.
- 정의 방법: {값1, 값2, 값3, ....} 
    - 중괄호에 값 넣기 (딕셔너리는 키와 값이 들어가니까 차이가 있음!)
#### 집합연산
합집합: a | b-> a와 b의 합집합 (a.union(b))
교집합: a & b -> a.intersection(b)
차집합: a - b -> a.difference(b)
대칭차집합: a ^ b -> a.symmetric_difference(b)

#### 집합 자료로 형변환
set(리스트 또는 튜플)

```python
a = {10,10,10,20,30}
```

```python
a[0]
```

```python
a = {10, 20, 30, 40}
b = {30, 40, 50, 60}

print("합집합:", a|b)
print("교집합:", a&b)
print("차집합:", a-b)
print("대칭차집합:", a^b)
```
