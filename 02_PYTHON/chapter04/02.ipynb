#### 횟수가 정해져 있는 반복

```
for num in range(반복횟수):
    # 반복 사용되는 코드
```

i : 순서를 의미하는 index의 첫글자, 관례, 다음for문은 알파벳 순서대로 j,k,l,...
- range(반복 횟수), 0~반복횟수
- range(시작, 종료): 시작 이상, 종료 미만, range(2,5) -> 2,3,4
- range(시작,종료,증감단위)

#### 반복가능한 요소(iterable)의 반복
```
for 요소 in 반복가능 요소들:
    # 반복 수행되는 코드
```


````python
# 참고 - range() 숫자 생성시에도 사용
nums = [*range(1,101)]
nums[:5]
````

````python
total = 0
start = 1

for num in range(100): # 100번 반복
    # total = total + start
    total += start
    start += 1
    print("반복 횟수:", num)

print (f"1부터 100까지 함계: {total}")
````

### 2. while 문
- 참인 조건에서 반복을 수행
```
while 판별/조건식:
    조건식이 참일때 반복 수행되는 코드
```

````python
total = 0
start = 1

while start <= 100:
    total += start
    start += 1

print("함계:", total)
````

### 3. 반복문의 중간 제어: break와 continue
- continue: 현재 반복을 중단하고 새로운 반복을 시작(건너띄기)
- break: 반복 중단

````python
total = 0
start = 1

for i in range(1,101): # 1~100
    if i % 2 == 1: # 홀수인지 판별
        total += i

print(f"합계:{total}")
````

````python
total = 0
start = 1

for i in range(1,101): # 1~100
    if i % 2 == 0: # 짝수
        continue # 현재 반복문의 코드 실행을 중단하고 다시 반복 -> 건너띄기
    total += i

print(f"합계:{total}")
````

````python
for _ in range(10):
    num = int(input())

    if num == 99:
        print("종료")
        break # 반복 종료
    print("입력한 숫자:", num)
````

### 4. 이중 반복문과 시퀀스 중첩 순회

대표적인 얘시: 구구단
- 첫번째 반복은 단(2~8), 8회 반복
- 각 단마다 1~9를 곱하는 반복


````python
for i in range(2, 10): # 단(2~9단)
    print("-"*5, i, "단", "-"*5)
    for j in range(1, 10): # 곱하는 수
        print(f"{i} x {j} = {i*j}")
````
