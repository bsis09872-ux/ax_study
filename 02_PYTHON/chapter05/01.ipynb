            - 함수도 주소가 있음. 예: calc()함수는 calc라는 이름이 해당 함수의 주소를 참조하고 있는 것.

````python
def add (num1, num2):
    num3 = 30

    result = num1 + num2 + num3 # 지역변수 result

    return result
````

````python
result = add(10,20) # 전역 변수 result

result
````

- 위 함수 내부의 result 변수와 아래 외부에 있는 result 변수는 서로 다른 변수이다.
- 위는 함수 내에서 정의된 지역변수이고 아래는 전체에서 정의된 전역 변수이다.
- 즉 주소, 저장된 메모리 위치가 다른 변수들이다.

자료구조 (추가)
스택(stack) vs. 큐(Queue)
- 스택 구조: 아래가 막힌 구조 (통?)
    - 가장 나중에 넣은 데이터를 가장 먼저 꺼내는 후입선출(LIFO, Last In First Out) 방식의 자료구조입니다. (넣어서 쌓은 순서대로 거꾸로 하나씩 꺼내서 써야하는 구조임)
- 큐 구조: 대기열
    - 먼저 들어온 데이터가 먼저 나가는 선입선출(FIFO, First In First Out) 방식의 선형 자료구조입니다. 
    - 데이터를 넣는 enqueue와 빼는 dequeue가 양 끝(front와 rear)에서 따로 이루어집니다. 은행 대기표나 줄서기와 같습니다

````python
# 함수 안에 함수가 정의된 상황

def func(num):
    result = add(10,20)

    result += num

    return result
````

````python
result = func(40) # 실행순서: func() -> add()

result
````

- call stack (스택 호출)
- 함수 호출 순서
    (1) func(40)
        PyFrameObject(func) 
        f_locals num = 40, reesult = 100
    (2) add (10,20)
        PyFrameObject(add)
        f_locals: num1 = 10, num2 = 20

- 함수 사용 이후 제거 순서
    (1) add()
    (2) func()

    -> 내부 구조가 수정이 된 후 가장 바깥의 구조가 수정될 수 있음

### 2. 매개변수(parameter)와 인수(argument)
- 매개 변수: 함수 호출시 값을 담아주는 변수 (위 calc함 수 기준: x)
- 인수: 호출시 사용된 값 (위 calc함 수 기준: 10)

### 3. 위치 인수(positional arguments)와 키워드 인수(keyword arguments)


- 위치 인수: 매개변수의 순서대로 값을 지정
- 키워드 인수: 매개 변수 명과 함께 값을 지정, 순서는 중요하지 않음
- 위치인수 먼저 사용 후 키워드 인수 사용

````python
def add(num1, num2):
    result = num1 + num2
    # return num1+ num2 (이렇게 한줄로 정리해도됨; 변수가 2개로 좀 더 간단한 구조)
    return result
````

````python
add(10,20) # 위치 인수
````

````python
add(num1=18, num2 = 20) # 키워드 인수
````

````python
add(num2= 20,num1=10) #키워드 이름이 중요하지 순서는 중요하지 x
````

````python
def add(num1, num2, name, mobile):
    return f"{num1 + num2}, {name=}, {mobile=}"
````

````python
add(10,20,mobile = "010-0000-0000", name= "김영희")
````

### 5. 가변인수(*args)와 키워드 가변 인수(**kwargs)
- *args: 가변적인 위치 인수, 매개변수 갯수 상관없이 지정 가능, 키워드 매개변수로는 사용 불가

````python
def add (num1, num2, num3, num4):
    return num1 + num2 + num3 + num4

add (10,20,30,40)
````

````python
def add(*nums):
    # print(nums, type(nums))
    total = 0
    for num in nums:
        total += num

    return total
````

````python
print(add(10,20))
print(add(10,20,30))
````

````python
def print_user_info(**info):
    print(info)
````

````python
print_user_info(name= '김철수', age=20)
````

````python
def print_user_info(name, age, **info):
    print("name:", name)
    print("age:", age)
    print(info)
````

````python
print_user_info("김철수", 20, mobile="010-0000-0000")
````

````python
def print_user_info(name, age, *args, **kwargs):
    print("name:", name)
    print("age:", age)
    print(args)
    print(kwargs)
````

````python
print_user_info("김철수", 20, 'A', 20, mobile="010-0000-0000")
````

````python
import sys

def outer_func():
    x = 10

    def inner_func():
        y = 20

        # 1. 현재 실행 중인 프레임 객체 가져오기
        frame = sys._getframe()

        print(f"1. 현재 실행 함수: {frame.f_code.co_name}")
        print(f"2. 현재 지역 변수(f_locals): {frame.f_locals}")
        print(f"3. 전역 스코프 참조 여부: {'outer_func' in frame.f_globals}")

        # 2. f_back을 통해 부모 프레임(outer_func)으로 이동
        parent_frame = frame.f_back
        print(f"4. 부모 함수 이름: {parent_frame.f_code.co_name}")
        print(f"5. 부모 함수의 지역 변수: {parent_frame.f_locals}")

    inner_func()

outer_func()
````
