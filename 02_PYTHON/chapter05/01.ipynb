## 5-1: 함수 정의와 매개변수

### 1. 함수의 기본 정의와 작동 방식
#### 정의 방법
```
def 함수이름(매개변수, ...):
    # 처리 부분

    return 반환값 # 출력 부분
```

#### 호출 방법 (실행)
함수이름(...) -> 엔터

#### 목적
코드의 재활용

````python
def calc(x):
    y = x * 2 + 1

    return y
````

````python
result = calc(10)

result
````

### 2. 매개변수(parameter)와 인수(argument)
- 매개 변수: 함수 호출시 값을 담아주는 변수 (위 calc함 수 기준: x)
- 인수: 호출시 사용된 값 (위 calc함 수 기준: 10)
