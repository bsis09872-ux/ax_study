## 5-2: 변수 스코프와 람다 함수

- 변수 스코프(Scope) : 변수를 접근할 수 있는 범위

### 1. 전역 변수(global)와 지역 변수(local)
- 전역변수(global): 함수 외부에 정의 변수, 코드 전역에서 접근 가능
- 지역변수(local): 함수 내부에 정의된 변수, 함수가 실행될때만 유효한 변수


```python
a = 10 # global var

def func():
    b = 20 # local var
    print(f"{a=},{b=}")

# print(a,b)
print(a)

func()
```

```python
c = 30

def outer():
    a = 10

    # print("b:", b)
    print("c:",c)

    def inner():
        b = 20
        print(f"{a=},{b=}{c=}")

    inner()

outer()
```

### 2. 파이썬의 변수 검색 규칙: LEGB 스코프

접근 우선순위: 좁은 범위가 우선순위가 높다!

- L(local - 지역변수) > E(enclosing > 외부 함수 변수) > G(global - 전역변수) > B(builtin - 내장 함수)



```python
a = 'global'

def outer():
    a = 'enclosing'
    print("outer:", a)

    def inner():
        a = 'local'
        print("inner:", a)

    inner()

outer()
```

### 3. global 키워드

```python
a = 'global'

def outer():
    a = 'enclosing'
    print("outer:", a)

    def inner():
        global a # a라는 변수는 전역변수를 찾겠다.
        
        print("inner:", a)

    inner()

outer()
```

```python
a = True

def outer():
    a = 'enclosing'
    print("outer:", a)

    def inner():
        global a # a라는 변수는 전역변수를 찾겠다.
        a = not a
        print("inner:", a)

    inner()

outer()
```

### 5. lambda 함수
- 간단하게 일회용으로 사용하는 경우 한줄의 식으로 간단하게 함수 정의
- 일회용으로 사용되므로 매개변수 이름도 최대한 간결하게 (한 글자)


```python
square = lambda num : num ** 2

square(10)
```

```python
def change_numbers(callback, *args):
    numbers = list(args)

    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])

    return numbers
```

```python
change_numbers(lambda num: num **2, 10, 20, 30)
```

### 7. lambda와 map(), filter()를 활용한 데이터 일괄 가공
- map(): 변환 작업
- filter(): 특정 조건에서 데이터를 선별

map() &rarr; map(function, iterable, *iterables)

- 매개변수에 함수, 이터러블: 리스트나 튜플, 위치기반 가변인수가 있음

```python
fruits = ['사과','오렌지', '망고']
list(map(lambda x : f"**{x}**", fruits))
```

filter() &rarr; filter(function, iterable, /)



```python
numbers = [*range(1,11)]

list(filter(lambda x: x % 2 ==1, numbers))
```
