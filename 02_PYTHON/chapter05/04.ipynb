## 5-4: 이터레이터와 제너레이터
1. 이터러블(Iterable)
- 반복 가능한 객체
- \_\_iter\_\_(...) 정의되어 있다ㅏ면 반복 가능 !
-\_\_next\_\_(...) 인덱스를 한칸씩 이동하면서 요소를 꺼내오는 기능을 정의
- 이터레이터(반복자)

````python
nums = [10,20,30]

nums.__iter__
````

````python
for num in nums:
    print(num)
````

2. 이터레이터(Iterator)
- 반복을 하려면 이터레이터 객체 생성 (__iter__)
    > 호출: iter(...)
- 순서를 한칸씩 이동하면서 요소 조회(__next__)
    > 호출: next(...)

````python
fruits = ['apple', 'orange', 'melon', 'banana']

it = iter(fruits) #__iter__(...) 호출
it
````

````python
next(it)
````

3. 제너레이터(generator)
- 함수의 실행을 정지하고 원할 때 다시 재개하는 형태

````python
def generate_nums():
    print('1번째')
    yield 10 # 분
    print('2번째')
    yield 20
    print('3번째')
    yield 30
````

````python
gen = generate_nums()
gen
````

````python
next(gen)
````

4. 제너레이터 표현식
- 참고: 리스트 컴프리헨션 -> 생성 결과가 처음부터 메모리에 적재
- 리스트 대신 튜플로 리스트 컴프리헨션과 동일하게 사용하면 된다. -> 메모리 소비가 적다.
- 대용량 데이터 처리

````python
gen = (i for i in range(1,101) if i % 2 == 0)

gen
````

````python
next(gen)
````

````python
for num in gen:
    print(num)
````
