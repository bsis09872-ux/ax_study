```python
counter = 0

def increase():
    global counter
    counter += 1 #전역변수의 수치를 1씩 증가

def local_increase():
    counter = 100

increase()
increase()
local_increase()
```

```python
numbers = [1,2,3,4,5]

result = list(filter(lambda x: x % 2 ==0, map (lambda x: x + 1, numbers)))
```

#### 역할의 차이 📌
1. map(): 모든 요소를 변환(가공)하는 도구
> "모든 숫자에 1을 더해줄래"
2. filter(): 조건에 맞는 것만 선별(필터링)하는 도구
> "짝수인 것만 골라줄래"
