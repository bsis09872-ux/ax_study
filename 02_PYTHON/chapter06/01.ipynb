## 6-1: 모듈과 패키지

1. 독립 소스 파일 모듈 (Module)
- 모듈 - 파이썬 파일 하나
- 다른 파일에서 모듈을 불러와서 사용하는 경우가 많음
- 모듈명 = 파일 이름
- __name__: 모듈 이름

```python
import mod1
```

2. 디렉터리형 패키지(package)
- 패키지는 모듈을 모아 놓은 디렉토리
- from 패키지명.하위패키지명... import ...
- 전체 패키지 경로: 절대 경로 접근 예) chapter.mod
- 상대 경로 접근법(..) : 바로 위 패키지
-               (.): 현재 모듈이 있는 패키지

3. import, from, as 키워드
- import : 모듈 불러오기
- 불러온 모듈에서 마침표(.)를 통해서 접근
- from 모듈명/패키지명 import 모듈에 있는 함수, 변수, 클래스 등등
- from 모듈명 import * -> 모듈의 전체 함수, 변수, 클래스를 불러온다.
- as: alias - 별칭

```python
import mod1 
```

```python
mod1.add(10,20)
```

```python
mod1.VERSION
```

```python
import mod1 as m

m.add(10,20)
```

```python
from mod1 import add, minus, VERSION
```

```python
add(10,20)
```

```python
VERSION
```

```python
from mod1 import *
add(10,20)
```

#### __init__.py
- 현재 디렉토리가 패키지임을 알려주는 파일
- 하위 모듈, 하위 패키지에서 공통적으로 처리해야 하는 코드를 작성, 모듈을 불러올때 최초 한번만 실행
- 패키지내에서 사용할  공통적인 함수, 변수 클래스를 정의
- from 패키지 import * : __all__ = [...]: 에 정의된 모듈만 임포트

```python
from game import lol
```
