3. try-except-else-finally

예외 처리 구문
```
try:

    # 예외가 발생할 가능성이 있는 코드

except:

    # 예외 발생 시 유입되는 구간, 대안 처리
    # 대안 로직, 로깅 처리

else:
    예외가 발생하지 않아야 유입되는 구간
finally: 
    
    # 예외가 발생하든 정산 처리되든 항상 유입되는 구간, 공통적인 처리
    # 자원 정리 (메모리 해제)

```

- 모든 예외 클래스는 Exception의 자식 클래스
- 특정 예외를 특정하지 못하는 경우 exception으로 처리 가능


````python
try:
    num = int(input())
    result = 10 / num

    print(result)

except:
    print("0이 아닌 정수를 입력하시오.")

else: 
    print("정상처리")
finally:
     print("logging...")
    
````

5. 다중 예외 처리 (try-except-else-finally)

````python
try:
    num = int(input())
    result = 10 / num

    print(result)

except ZeroDivisionError :
    print("0이 아닌 정수를 입력하시오.")
````

````python
try:
    num = int(input())
    result = 10 / num

    print(result)

except ZeroDivisionError :
    print("0이 아닌 정수를 입력하시오.")
except ValueError:
    print ("정수만 입력하세요.")
````

````python
try:
    num = int(input())
    result = 10 / num

    print(result)

except (ZeroDivisionError, ValueError) :
    print("0이 아닌 정수를 입력하시오.")
````

````python
try:
    num = int(input())
    result = 10 / num

    print(result)

except ZeroDivisionError as e :
    print(e)
````

````python
try:
    num = int(input())
    result = 10 / num

    print(result)

except ZeroDivisionError:
    print("0이 아닌 정수를 입력하시오.")
except Exception as e: # exception은 가장 하단쪽 except로 정의
    print(e)
````

6. raise 활용한 데이터 유효성 검증

- 예외 발생 시키기
```
raise 예외 클래스("메세지...")
```

````python
def login(user_id, user_pass):
    if user_id != 'user01':
        print("아이디가 일치하지 않습니다.")
        return

    if user_pass != "1234":
        print("비밀번호가 일치하지 않습니다.")
        return

    print("로그인 처리 로직 ...")
````

````python
def login(user_id, user_pass):
    if user_id != "user01":
        raise ValueError("아이디가 일치하지 않습니다.") #예외가 발생, 코드 실행 흐름은 중단

    if user_pass != "1234":
        raise ValueError("비밀번호가 일지하지 않습니다.")

    print("고르인 처리 로직 ...")
````

````python
# 회원페이지
try: 
    login("user02", "1234")
except ValueError as e:
    print(e)
    print("회원 페이지에서 추가적으로 할 로직...")
````

````python
# 관리자페이지
try: 
    login("user02", "1234")
except ValueError as e:
    print(e)
    print("관리자 페이지에서 추가적으로 할 로직...")
````

7. 사용자 정의 예외
- Exception은 모든 예외의 부모클래스
- Exception을 상속 받으면 또는 Exception을 상속받은 자식 클래스를 상속 받으면 그 클래스도 예외 클래스!

````python
class UserIdMissmatchError(Exception):

    def __str__(self):
        return "아이디가 일치하지 않습니다."

class UserPassMissmatchError(Exception):

    def __str__(self):
            return "비밀번호가 일치하지 않습니다."

````

````python
# def login(user_id, user_pass):
#     if user_id != "user01":
#         raise UserIdMissmatchError() #예외가 발생, 코드 실행 흐름은 중단

#     if user_pass != "1234":
#         raise UserPassMissmatchError()
#     print("고르인 처리 로직 ...")

def login(user_id, user_pass):
    if user_id != "user01":
        raise UserIdMissmatchError # 예외 상황에 대해서는 ()괄호로 실행을하지 않아도 돌아간다.

    if user_pass != "1234":
        raise UserPassMissmatchError
    print("고르인 처리 로직 ...")

````

````python
# try:

# except UserIdMissmatchError as e:

# except UserPassMissmatchError as e:

try:
    login("user01", "12345")

except(UserIdMissmatchError, UserPassMissmatchError) as e:
    print(e)

````

8. traceback
- traceback.print_exc(): 에러 스택 콘솔 출려
- traceback.format_exc(): 에러 스택을 문자열로 반환
- traceback.print_stack(): 에러 없이 현재 호출 흐름 확인

````python
import traceback

try:
    result = 10 / 0
    print(result)

except: 
    print("에러 발생")
    # traceback.print_exc()
    message = traceback.format_exc()
    print("message", message)
````
