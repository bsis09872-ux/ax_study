## 1. 클래스 변수 (정적 변수), 인스턴스 변수

```python
class User:
    pass

user = User()
user.name = "김철수"
user.age = 20

print(user.name, user.age)
```

```python
class User:
    count = 0
```

```python
u1 = User()
u2 = User()

print(u1.count, u2.count)
```

```python
u1.count = 10

print(u1.count, u2.count)
```

## 2. @classmethod, @staticmethod

- 객체를 생성하지 않아도 접근할 수 있는 매서드를 정의할때

```python
class Calculator:
    count = 0 

    # clz -> calculator 클래스 객체 참조 변수
    # clz -> 클래스 변수, 다른 클래스 메서드를 접근하기 위한 목적
    @classmethod
    def add(clz, num1, num2):
        clz.count += 1
        print("호출횟수:", clz.count)
        return num1 + num2

    @staticmethod
    def minus(num1, num2):
        return num1 - num2
```

```python
Calculator.add(10,20)
```

```python
Calculator.minus(10,20)
```
