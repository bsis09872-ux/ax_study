4. mean(), median()
- statistics 모듈에 포함된 표준 라이브러리
- mean(iterable): 평균값
- median(iterable): 중앙값

```python
import statistics 

nums = [10,15,20,8,7,10000]
print("평균:", statistics.mean(nums),"중앙값:", statistics.median(nums) )
```

5. map(), filter()
- map():변환
- filter(): 판별 &rarr; 정제

    &rarr; 리스트 컴프리헨션으로 대체 가능 !!!

```python
nums = [*range(1,11)]

nums2 = [x**2 for x in nums]
print(nums2)

nums3 = [x for x in nums if x % 2 ==1]
print(nums3)
```

6. enumerate()

```python
fruits = ['사과', '오렌지', '멜론', '망고']

num = 1
for fruit in fruits:
    print(num, fruit)
    num += 1
```

```python
for item in enumerate(fruits): # (순서 번호, 요소)
    idx = item[0]
    fruit[1]
    print(idx, fruit)
```

```python
for i, fruit in enumerate(fruits): # (순서 번호, 요소)
    
    print(i, fruit)
```

```python
for i, fruit in enumerate(fruits, start =1 ): # (순서 번호, 요소)
    
    print(i, fruit)
```

7. zip()

```python
persons = ['김철수', '김영희', '김민수']
fruits = ['사과', '오렌지', '멜론']
```

```python
items = []
for i, person in enumerate(persons):
    items.append((person,fruits[i]))

items
```

```python
items = zip(persons, fruits)

list(items)
```

8. sorted()
- 리스트.sort(): 원본 데이터 정렬
- sorted(Iterable): 원본 데이터 유지, 새로운 객체를 생성해서 정렬 (불변객체)

```python
import random
nums = [*range(1, 11)]
random.shuffle(nums)

nums
```

```python
nums.sort()

nums
```

```python
nums_sorted = sorted(nums)

print(nums_sorted)
print(nums)
```
