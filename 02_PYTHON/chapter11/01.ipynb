# 동기 vs 비동기 이해

```python
# 동기식 함수 (서브 루틴)

import time

def work(title, seconds):
    time.sleep(seconds)
    print(title, "완료!")

start = time.perf_counter() # 작업 시작 시각

work("작업A", 2)
work("작업B", 3)
work("작업C", 1)

end = time.perf_counter() #작업 종료 시각

print("걸린시간:", end - start)
```

```python
# 비동기 함수 (코루틴)
import asyncio
import time

async def async_work(title, seconds):

    asyncio.sleep(seconds)
    print(title, "완료!")



async def main():
    start = time.perf_counter()
    await asyncio.gather(
    async_work("작업A",2),
    async_work("B작업",3),
    async_work("C작업",1)

    )   
    
    end = time.perf_counter()
    print("걸린 시간:", end-start)

await main()
```
