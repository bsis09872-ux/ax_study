### 8장 1강 실습 3번 (심화1)

```python
class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100

    def take_damage(self, damage):

        minus_damage = damage if self.hp >= damage else self.hp
        self.hp -= minus_damage
        print("남은 체력:", self.hp)

        if self.hp <= 0:
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")
```

```python
hero = Hero("영웅")
hero.take_damage(90)
hero.take_damage(20)
```
