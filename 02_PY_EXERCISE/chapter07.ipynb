# 7장 1강 예외 처리

```python
# 1 정수 변환 예외 처리

def guess_age(yr):
    age = 2026 - yr + 1
    intl_age = 2026 - yr
    print(f"당신의 올해 나이는 {age}(만 {intl_age}세) 입니다.")


try:
    yr = int(input("태어난 연도를 입력하시오: "))
except ValueError as e:
    print(f"경고: {e} - 숫자만 입력할 수 있습니다.")
       
```

```python
# 2 메뉴 선택 및 다중 얘외 처리하기

menu = ['고구마케이크', '잠봉뵈르', '망고빙수']

try:
    menu_number = int(input("원하는 메뉴의 인덱스 번호: "))
    print(menu[menu_number])  # ← 여기서 IndexError 발생 가능
except ValueError as e:
    print(f"경고: {e} - 숫자만 입력 가능합니다.")
except IndexError as e:
    print(f"경고: {e} - 0부터 2까지 숫자 중 입력해주세요.")
else:
    print("메뉴가 선택되었습니다.")
finally:
    print("종료합니다.")
```

```python
# 3 raise 키워드를 활용한 데이터 유효성 검증

def validate_product_count(product_count):
    if product_count <= 0:
        raise ValueError ("주문 수량을 확인하시오.")
    else:
        print(f"상품 수량: {product_count}\n상품의 수량이 정상적으로 등록되었습니다.")

try:
    validate_product_count(product_count=int(input()))
except ValueError as e:
    print(f"[경고] {e} - 숫자 1 이상의 정수로 입력하시오.")



```

# 7장 2강 디버깅 기초 (1)

```python
# 1 에러 트레이스백 역추적 및 원인 진단
def calculate_share(total_price, num_people):
    share_per_person = total_price / num_people
    return share_per_person

def start_settlement():
    participants = 0
    result = calculate_share(50000, participants)
    print(result)

start_settlement()
```

```python
# 2 VS Code 디버거와 중단점을 활용한 실행 일시 정지
def calculate_revenue():
    total_revenue = 0
    daily_sales = [12000, 25000, 31000]
    for amount in daily_sales:
        total_revenue += amount
    print(total_revenue)

calculate_revenue()
```

```python
# 3 Step Over 명령을 통한 단계별 실행 흐름 제어
def sum_scores():
    final_score = 50
    bonus_scores = [10, 20, 30]
    for score in bonus_scores:
        final_score += score
    return final_score

result = sum_scores()
```
