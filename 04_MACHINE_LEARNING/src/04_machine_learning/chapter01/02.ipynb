#### 4.1 데이터 로드 및 결측치, 이상치 처리

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("titanic")
```

결측치 파악하기

```python
df.isna().sum()
```

숫자형 데이터(나이): 전체 탑승객의 '평균(mean)'으로 빈칸 채우기

```python
mean_age = df['age'].mean()
print("평균 나이:", mean_age)

df['age'] = df['age'].fillna(mean_age)

df.isna().sum()

# 만약 age가 중요한 지표인 경우에는 누락시키는 것이 좋다. 
# 이렇게 대치하는 것은 크게 중요하지 않은 지표일때 표본 크기를 유지하기 위해 처리한다.
```

문자/범주형 데이터(탑승 항구): 가장 많이 탑승한 곳인 '최빈값(mode)'으로 빈칸 채우기

```python
mode_embarked = df['embark_town'].mode()[0] # 빈도수가 동일하면 최빈값이 여러개 나올 수 있어서 첫번째 값만 인덱싱해옴

mode_embarked
```

```python
df['embark_town'] = df['embark_town'].fillna(mode_embarked)
df['embarked'] = df['embark_town'].map(lambda x: x[0].upper)

df.isna().sum()
```

특정 컬럼(deck) 을 '기준'으로 좌우(열 방향) 데이터를 채우기

```python
df['deck'] = df['deck'].ffill()

print(df.isna().sum())

df.head()
```

```python
df['deck'] = df['deck'].bfill()

print(df.isna().sum())
```

너무 비싼 요금(이상치) 걸러내기

```python
print("요금 평균:", df['fare'].mean())
print("중앙값:", df['fare'].median())
```

```python
plt.figure(figsize=(4,4))
plt.hist(df['fare'], bins=10)
plt.show()
```

결측치/이상치 보완 결과 확인

```python
df = df[df['fare']<=300]
df.isna().sum()
```

```python
df[df['fare'] > 300].count()
```

#### 4.2 중복 데이터 처리하기

중복 데이터 확인하기

- duplicated() : 중복 행을 확인

```python
# 중복 값은 True로 나옴 -> 합계를 구할 수 있음

df.duplicated().sum()
```

중복 데이터 제거하기

- drop_duplictes() : 중복 행 삭제

```python
# df.drop_duplicates()[:10]
df = df.drop_duplicates().reset_index(drop=True)
print("중복 데이터 수:", df.duplicated().sum())
```

#### 4.3 그래프로 시각화하기

내 컴퓨터에 맞는 한글 폰트 설정하기

```python
import platform
from matplotlib import rc

os_name = platform.system()
if os_name == "Windows":
    rc("font", family="Malgun Gothic")
else:
    rc("font", family="AppleGothic")

```

도화지(Figure)와 3개의 액자(Axes) 준비

```python
fig, axes = plt.subplots(1,3, figsize=(18,5))
fig.suptitle("타이타닉 탑승객 데이터 분석", fontsize= 16)

# [히스토그램] 탑승객 나이 분포
axes[0].hist(df['age'], bins=20, color='skyblue', edgecolor='black')
axes[0].set_title("탑승객의 나이 분포")
axes[0].set_xlabel('나이')
axes[0].set_ylabel('승객수')

# [박스 플롯] 티켓 등급에 따른 요금 이상치 확인
sns.boxplot(x='class', y='fare', data=df, ax=axes[1])
axes[1].set_title('티켓 등급별 요금 분포')
axes[1].set_xlabel('티켓 등급')
axes[1].set_ylabel('티켓 요금')


# [산점도] 나이와 클리닝된 요금의 관계
scatter = axes[2].scatter(
    df['age'],
    df['fare'],
    c=df['survived'], # 1 - 빨강, 0 - 파랑
    cmap='coolwarm', # 테마
    s=df['fare'] / 2, # 점 크기 분류 - 2개로 나눔
    alpha=0.7 # 점 투명도 값
)

fig.colorbar(scatter, ax=axes[2], label = "생존여부")
```
