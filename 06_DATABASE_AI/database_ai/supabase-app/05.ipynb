## 메서드 체이닝
- 선언적 방식 (execute실행 전까지는 실행되는 것이 아님. 즉 앞의 쿼리들은 모두 선언형식/ 선언문 안에서는 동일한 객체를 매개체로 받는다 -  내부 봔환값)
```
supabase.table(...)
    .select(...)
    .eq(...)
    .execute()
```


````python
class SupabaseEx:
    def __init__(self):
        self.settings = []

    def table(self,table):
        self.settings.append(table)

        return self

    def select(self, fields):
        self.settings.append(fields)

        return self #생성된 동일 객체 (반환)

    def eq(self, name, value):
        self.settings.append(f"{name}={value}")

        return self

    def execute(self):
        print(self.settings)
        print("실행...")
        return "처리된 결과..."
````

````python
supabase = SupabaseEx()

response = (
    supabase.table("users")
        .select("*")
        .eq("email", "user01@test.org")
        .execute()
)

response
````

````python
class Product:

    # 내부 식별자 (_sentinel = token 확인?)
    _sentinel = object()

    # 클래스 외부에서 직접 생성은 차단, 
    def __init__ (self, token, name, price, description):

        if token is not Product._sentinel: #외부에서 직접 생성한 경우
            raise Exception("직접 생성할 수 없습니다.")

        self.__name = name
        self.__price = price
        self.__description = description

    def name(self):
        return self.__name

    def price(self):
        return self.__price

    def description(self):
        return self.__description

    def __repr__(self):
    

    # Product 객체를 조립하고 생성하는 클래스
    class Builder:
        def name(self, name):
            self.__name = name

            return self

        def price(self, price):
            self.__price = price

            return self

        def description(self, description):
            self.__description = description

            return self

        # 객체를 생성
        def build(self):
            return Product(Product._sentinel, self.__name, self.__price, self.__description)


````

````python
product = (
     Product.Builder()
        .name("아이폰")
        .price(10000)
        .description("설명")
        .build()
)

product
````
