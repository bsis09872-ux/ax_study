### 2. 초기 자바스크립트의 변수 선언: var의 치명적인 결함

```

```

`번역 과정`을 거쳐서... `window 객체의 하위 속성`으로 추가된다. (웹브라우저 기준) / `global 객체의 하위 속성`으로 추가 (Ndoe.js 기준)

#### 2.1 var의 3가지 결함
- 호이스팅 (Hoisting)
- 선언 키워드 생략 허용 (암묵적 전역 변수)
- 함수 레벨 스코프 (Function-level Scope)

````javascript
var i = 100000; // window.i /global.i = 100000

console.log(i);

for (var i = 0; i < 10; i++) { // window.i / global.i = 0~9
    //반복로직
    console.log("for문에 정의된 i:", i);
}

console.log(i); //window.i / global.i = 10
````

#### 4.1 자바스크립트의 주요 데이터 타입

- 조건식 : 참/거짓을 판별하는 연산식 (비교, 논리 연산 주로 쓰임)
```

    //조건식 참일때 실행되는 구간

```

```
if (조건) {
    //참일때 실행되는 구간
} else if (조건) {
    //참일때 실행되는 구간 2
} else {
    // 모든 조건이 거짓일때 실행되는 구간

```

- 삼항 조건식 (메모: 파이썬의 람다함수 같이 한줄로 간단히 사용하는 js의 조건문)
- 변수 = (조건식) ? 참일때 실행되는 코드 값 : 거짓일때 실행되는 코드나 값;
- 조건이  하나일때는 소괄호 안써도 작동됨

````javascript
let num = 10;

const result = num == 10 ? "숫자 10입니다." : "숫자 10이 아닙니다."

console.log(result)
````

````javascript
let greeting;

greeting = greeting || "기본값";

console.log(greeting)
````

````javascript
let num = 0;

num = num || 10;

console.log(num);
````

````javascript
// 0을 false로 인식하는 문제 발생!
// ?? -> nullish 연산자 -> null, undefined 일때만 판단
// let num;
let num = null;

num = num ?? 10;
console.log(num);
````

- && : 단락회로평가를 이용한 간단한 조건식 수행
- AND연산자는 하나라도 값이 거짓이 나오면 조건문 수행이 종료됨. 

````javascript
const message = "";

message && console.log(message); // 앞의  항이 참일때 뒤의 항의 실행됨. 이걸 간단한 조건식으로 사용함!
````

#### 6.2 선택문 (switch - case)

- switch ~ case : case 키워드에 매칭되면 그게 바로 코드 실행 시작점이고 break를 만날때까지 실행된다.
- break : 실행흐름을 중단
- default : case에 매칭된 키워드가 없으면 마지막에 유입되는 구간

````javascript
const rank = 4;

switch(rank) {
    case 1: 
        console.log("금매달");
        break;
    case 2:
        console.log("은매달");
        break;
    case 3:
        console.log("동매달");
        break;
    default:
        console.log("등수는 1~3 중에서 입력!")
}

// typescript랑 같이 있어서 오류난듯 - 브라우저에서 실행함
````

### 7. 예외 처리 (try - catch - finally)

#### 7.1 try - catch - finally 문의 구조

```
try {
    // 예외가 발생할 가능성이 있는 코드
} catch(err) {
    // 예외가 발생하면 유입되는 구간 - 대안적인 처리, err: 발생한 예외에 대한 정보
} finally {
    // 예외 발생 여부와 무관하게 무조건 실행되는 구간
}
```
