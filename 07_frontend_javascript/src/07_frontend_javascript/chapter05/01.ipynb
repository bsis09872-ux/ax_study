```javascript
function add(num1, num2 = 10) {
    return num1 + num2;
}

add(20);
```

arguments 객체

```javascript
// 함수 호출시 사용한 인수 목록
function sum () {
    let total = 0;
    for (const arg of arguments) {
        total += arg;
    }

    return total;
}
```

```javascript
console.log(sum(10,20));
console.log(sum(10,20,30));
console.log(sum(10,20,30,40,50));
```

나머지 매개변수 (Rest Parameter, 가변 매개변수)

```javascript
// ES6+
// ... : 가변 매개변수
function sum (...params) {
    console.log(params);
}

sum(10,20,30,40);
```

```javascript
function sum (num1, ...params) {
    console.log(num1, params);
}

sum(10,20,30,40);
```

```javascript
// 전개 연산자
function add (num1, num2) {
    return num1 + num2;
}

const nums = [10,20];

console.log(add(nums[0],nums[1]));
console.log(add(...nums));
```

```javascript
const nums1 = [10,20,30,40];
// const nums2 = nums1; // 이거는 복제가 아니라 주소를 공유하게 되는것!
const nums2 = [...nums1]; // nums1 배열의 변수들을 차례로 가져와서 전개함; 결국 복사!

console.log(nums2, nums1===nums2);
```

#### 4.1 용어의 유래: 차별 없는 '동등한 권리’

#### 4.2 일급 객체로서의 3대 속성
- 변수 할당 (Assignment)
- 매개변수 전달 (Argument Passing) / 콜백 함수 / 사용자 정의 함수 / 일회용 함수
- 반환값 활용 (Return Value)

```javascript
function func1() {
    console.log("func1()");
}

function func2(f) {
    console.log("func2()");
    f();
}

func2(func1); 
```

```javascript
function square(num) {
    return num * num * num;
}

function calc(nums, callback) {
    let total = 0;
    for (const num of nums) {
        if (typeof callback === 'function') {
            total += callback(num);
        }
        
    }
    return total;
}
```

```javascript
calc([10,20,30], square);

typeof square;
```

```javascript
calc([10,20,30], function(num) {
    return num * num;
});
```

```javascript
calc([10,20,30], x => x * x);
```

### 5. 화살표 함수 (Arrow Function)

#### 5.1 정의 방법 및 스코프 생략 규칙

- function keyword 생략
- 매개변수 정의부와 코드 실행 영역({}) 사이에 => 로 표시
- 코드 구현 부분이 한줄이면 {} 생략 가능, return도 생략 해야 한다.
- 변수명도 한글자면 충분
- 매개변수가 1개면 () 생략 가능, 없거나, 2개 이상이면 안됨

```javascript
function square (num) {
    return num * num;
}
```

```javascript
const square2 = (num) => {
    return num* num;
};

square2(10);
```

```javascript
//const square3 = (num) => num * num;
//const square3 = (x) => x*x;
const square3 = x => x*x;
square(3);
```

#### 5.2 화살표 함수의 제한 사항
- 생성자 함수 사용 불가
- arguments 객체 미지원
- yield 사용 불가

#### 5.3 일반 함수 vs 화살표 함수의 this 바인딩 비교

- 일반 함수의 this : 함수가 호출되면 호출한 객체
- 화살표 함수의 this : 함수가 정의되는 시점에 이미 결정되어 있는 this로 정의된다. (전역 스코프 내에 존재 -> 이미 결정된 this는 바꾸기 힘들다.)

#### 6. 자바스크립트 실행 평가 과정과 클로저 (Closure)

```javascript
function add(x) {
    return function(y) {
        return function(z) {
            return x+y+z;
        };
    };
}
```

```javascript
add10 = add(10);
```

```javascript
add10;
```

```javascript
add10_20 = add10(20);
```

```javascript
add(10)(20)(30); // 고차 함수 (higher order function) - 함수의 함수를 반환하는 함수
```

```javascript
const add10 = add(10); //factory function <- 첫번째 값이 10으로 고정된 새로운 함수 생성
```

```javascript
function work() {
    console.log("작업1 실행...");

    return function() {
        console.log("work2 ...");

        return function() {
            console.log("작업 3... 완료")

        };
    };
};
```
