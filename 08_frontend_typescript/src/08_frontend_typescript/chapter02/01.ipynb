### 2. 배열(Array) 타입 정의하기

#### 2.1 단일 타입 배열 정의 방법

대괄호 표기법 (Type[])

```typescript
const orderPrices : number[] = [1000, 2000, 3000]; // number[] - 타입으로 선언된 배열 많이 사용!

orderPrices.push('ABC'); // 배열에 새로운 요소 추가 시도, 하지만 타입이 맞지 않으면 오류 발생
console.log(orderPrices);
```

제네릭 표기법 (Array)

```typescript
//const fruits : string[] = ['사과', '바나나', '포도']; // string[] - 타입으로 선언된 배열 많이 사용!
const fruits : Array<string> = ['사과', '오렌지', '바나나']; // Array<string> 다른 방법으로 배열 타입 선언 가능

console.log(fruits);
```

### 3. 고정 길이 데이터, 튜플(Tuple)

- 튜플(Tuple)이란?
> 배열 형식을 취하지만, 포함될 요소의 총 개수(길이)와 각 인덱스에 위치할 데이터 타입을 명확히 제한하는 구조입니다.

#### 3.1 튜플의 선언 구조

```typescript

const student : [string, number, boolean] = ['홍길동', 95, true]; // 튜플 타입 선언
// const student : [string, number, boolean] = ['홍길동', 95, true, 1]; // 튜플 타입 선언
console.log(student);

// 튜플 타입은 배열과 비슷하지만, 각 요소의 타입과 순서가 고정되어 있어요.
// 지정된 요소 이상을 추가하려고 하면 오류 발생 (Source has 4 element(s) but target allows only 3)
```

타입을 위반하거나 길이를 벗어나면 컴파일 오류가 발생합니다
