### 1. 유니온 타입 (Union Type): 다중 타입 허용하기

#### 1.1 유니온 타입의 기본 개념과 선언 문법
- 바(`|`) 기호를 기준으로 기재된 타입 중 하나를 만족하면 허용하는 타입 구조입니다.
- 숫자나 문자열 등 여러 타입의 데이터가 입력될 수 있는 변수를 정의할 때 사용합니다.

```typescript
let userId: string | number; 

userId = "user01";
userId = 10000;
console.log("User ID:", userId);
```

[추가 필기]
- 원시타입 객체에서는 이런 | 타입 교집합 연산자를 사용하지 않는다 
- let a: string & number; -> "never" 로 불가능함!
- 따라서 주로 OR 연산자인 | 파이프를 사용함. (둘 중 하나면 허용)

#### 1.2 객체 유니온과 공통 필드 접근 규칙
- 서로 다른 객체 타입들을 유니온으로 묶으면, 컴파일러는 모든 타입에 공통으로 존재하는 속성만 접근을 허용합니다. (접근 제한 원리)
- 특정 객체 타입에만 존재하는 속성에 접근할 때 발생할 수 있는 런타임 오류를 방지합니다. (안전성 보장)

```typescript
// 카드 결제 정보를 나타내는 인터페이스를 정의합니다.
interface CardPayment {
    amount: number; // 결제 금액 (공통 속성)
    cardNumber: string; // 카드 번호 (전용 속성)
}

// 계좌이체 결제 정보를 나타내는 인터페이스를 정의합니다.
interface BankTransfer {
    amount: number; // 결제 금액 (공통 속성)
    bankCode: string; // 은행 코드 (전용 속성)
}

type PaymentMethod = CardPayment | BankTransfer; //두 개의 결제 방식 중 하나

function processPayment(payment: PaymentMethod) {
    console.log("결제 요청 금액:", payment.amount);

    if ('cardNumber' in payment) { // CardPayment
        console.log("카드 번호:", payment.cardNumber);
    } else { //BankTransfer
        console.log("은행 코드:", payment.bankCode);
    }
    //console.log("카드 번호:", payment.cardNumber);
}

//duck type 정해진 특성이 하나라도 존재하면 해당 타입으로 분류함 (덕 타입핑). 다른 의미로 타입핑이 중요하다!

//카드 결제 객체
const myPayment: CardPayment = {
    amount: 50000,
    cardNumber: "1234-1234-1234-1234"
}

processPayment(myPayment) // Union type 사용했을경우 공통적인 것에만 접근이 가능하기에 어떤걸 골라야 할지 몰라서 존재하지 않는다고 에러 메세지가 뜸.
```

### 2. 인터섹션 타입 (Intersection Type): 복합 타입 합성하기

#### 2.1 인터섹션 타입의 기본 개념과 선언 문법
- 앰퍼샌드(`&`) 기호를 사용하여 지정된 모든 타입의 속성을 동시에 만족하도록 합성하는 타입입니다.
- 기존에 정의된 독립적인 타입들을 결합하여 확장된 타입을 생성할 때 활용합니다.

intersection 연산자는 interface extends에서는 사용하지 않음?

```typescript
// 사용자 기본 정보를 관리하는 타입을 정의합니다.
type UserProfile = {
  name: string; // 사용자 이름
  email: string; // 이메일 주소
};

// 시스템 권한 정보를 관리하는 타입을 정의합니다.
type UserPermissions = {
  adminLevel: number; // 관리자 등급
};

// 관리자 타입 (UserProfile + UserPermission)
// type SuperAdmin = UserProfile & UserPermissions;

// const adminUser: SuperAdmin = {
//   name: "user01",
//   email: "user01@test.org",
//   adminlevel: 1
// };

// console.log(adminUser);

type SuperAdmin = UserProfile & UserPermissions;
const adminUser: SuperAdmin = {
  name: "사용자01",
  email: "user01@test.org",
  adminLevel: 1,
};

console.log(adminUser);
```

#### 2.2 인터섹션 타입의 속성 확장과 주의사항
- 결합된 타입 내부의 속성을 갖추어야 컴파일 오류가 발생하지 않습니다.
- `string & number`와 같이 양립할 수 없는 원시 타입을 결합할 경우 어떤 값도 할당할 수 없는 `never` 타입이 됩니다.

```typescript
// 상품의 기본 정보를 정의하는 인터페이스입니다.
interface Product {
  id: number; // 상품 식별자
  name: string; // 상품명
  price: number; // 가격
}

// 재고 관리 정보를 정의하는 인터페이스입니다.
interface Inventory {
  stock: number; // 재고 수량
  location: string; // 창고 위치
}

```

### 3. 리터럴 타입 (Literal Type): 특정 값으로 범위 제한하기
리터럴 타입은 string이나 number 타입의 전체 범위 대신 지정된 값만을 허용하는 타입 정의 방식입니다.

#### 3.1 문자열 및 숫자 리터럴 타입
- 변수에 저장될 수 있는 값을 특정 문자열이나 숫자로 제한합니다.
- 정해지지 않은 문자열이나 숫자가 입력되는 상황을 컴파일 시점에 확인합니다.

```typescript
// 방향을 나타내는 문자열 리터럴 유니온 타입을 정의합니다.
type Direction = "UP" | "DOWN" | "LEFT" | "RIGHT";

//let dir: Direction = "DOWN";
let dir: Direction = "down";

// enum type도 비슷하게 사용 가능하지만 이제는 객체 상속이나 리터럴(고정값) 사용을 더 지향함.
```

#### 3.2 리터럴과 유니온·인터섹션을 결합한 종합 모델링

```typescript
// 주문 진행 상태를 지정하는 리터럴 유니온 타입입니다.
type OrderStatus = "PENDING" | "SHIPPED" | "DELIVERED";

// 주문 기본 데이터 구조를 정의합니다.
interface BaseOrder {
    orderId: string; // 주문 ID
    totalAmount: number; // 총 금액 
}

type DetailedOrder = BaseOrder & {
    status: OrderStatus
}
```
