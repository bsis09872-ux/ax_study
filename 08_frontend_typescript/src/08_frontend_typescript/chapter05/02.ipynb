### 5장 2강 : 타입 가드와 서로소 판별 유니온
타입 가드는 여러 타입이 섞인 데이터의 타입을 조건문으로 검증하여, 해당 조건문 블록 내부에서 컴파일러가 타입을 특정할 수 있도록 돕는 구문입니다.

#### 1.1 typeof 연산자를 활용한 원시 타입 좁히기
- 넓은 범위를 가진 유니온 타입을 특정 조건문 내부에서 더 구체적인 타입으로 제한하는 과정입니다.
- JavaScript 런타임에서 피연산자의 원시 타입 정보를 문자열로 반환하는 연산자입니다.

```typescript
// 다양한 타입의 입력값을 처리하는 함수 정의
function processInput(value: string | number) {
    if (typeof value === 'string') {
        console.log("문자열 길이:", value.length);
    } else {
        console.log("제곱근:", value*value);
    }
}

// 실습 실행
processInput("TypeScript");
processInput(20);
```

#### 1.2 in 연산자와 instanceof를 활용한 객체 타입 좁히기
- in 연산자: 객체 내부에 특정 속성(property) 이름이 존재하는지 확인합니다.
- instanceof 연산자: 객체가 특정 클래스의 인스턴스인지 확인합니다.

```typescript
// 관리자 및 일반 사용자 객체 타입 정의
type Admin = { name: string; privileges: string[] };
type User = { name: string; startDate: Date };
```

in 연산자를 활용한 객체 속성 검사 함수

```typescript

function printUserInfo(account: Admin | User) {
    console.log(`이름: ${account.name}`);
    if ('prvileges' in account) {// Admin type
        console.log(account.prvileges);
    }
}
```

instanceof 연산자를 활용한 클래스 객체 검사

```typescript
class Car {
    drive() {
        console.log("운전을 시작합니다.");
    }
}

class Truck {
    drive() {
        console.log("트럭을 운전합니다.");
    }
    loadCargo(amount: number) {
        console.log(`${amount}kg의 짐을 실었습니다.`);
    }
}

type Vehicle = Car | Truck;

function useVehicle(vehicle: Vehicle) {
    vehicle.drive();
    if (vehicle instanceof Truck) {
        vehicle.loadCargo(1000);
    }

    
}

// 실습 실행
const myTruck = new Truck();
useVehicle(myTruck);
```

### 2. 서로소 판별 유니온 (Discriminated Union)
서로소 판별 유니온은 객체 타입에 공통된 리터럴 태그 속성을 부여하여 조건문에서 타입을 명확하게 구분할 수 있도록 만드는 설계 패턴입니다.

#### 2.1 판별자(Discriminant) 태그를 통한 상태 모델링
- 판별자(Discriminant): 객체 타입을 구분하는 공통 속성이며, 서로 중복되지 않는 리터럴 값을 가집니다.
- 상태 구조화: API 요청 상태(로딩, 성공, 실패)와 같이 서로 독립적인 상황에 필요한 데이터를 정리할 수 있습니다.

```typescript
// API 통신 상태별 객체 타입 정의 (공통 속성 'status' 부여)
type LoadingState = {
    status: "loading";  // 리터럴 타입 태그
};

type SuccessState = {
    status: "success"; // 리터럴 타입 태그
    data: string[];
};

type ErrorState = {
    status: "error"; // 리터럴 타입 태그
    errorMessage: string;
};

// 세 가지 상태의 유니온 타입 생성 // status 공통적인 것의 값을 기준으로 구분함
type NetworkState = LoadingState | SuccessState | ErrorState;
```

#### 2.2 switch 문을 활용한 상태 분기

```typescript
// NetworkState에 따라 처리 내용을 결정하는 함수
function renderNetworkState(state: NetworkState) {
    // status 속성을 기준으로 타입을 판별
    switch (state.status) {
        case "loading": //LoadingStaute 타입
            console.log("LoadingState");
            break;

        case "suCcess": //SuccessState 타입
            console.log("SuccessState");
            break;

        case "error": //ErrorState 타입
            console.log("ErrorState:", state.errorMessage);
            break;
    }
    
}
// 요지는 힌트를 주어야 모두 사용 가능하고, 타입 좁히기를 하지 않으면 공통 부분만 접근이 가능하다는 것이다.

// 실습 실행
const responseSuccess: SuccessState = {
    status: "success",
    data: ["상품 A", "상품 B", "상품 C"]
};

const responseError: ErrorState = {
    status: "error",
    errorMessage: "네트워크 연결 오류"
};

renderNetworkState(responseSuccess);
renderNetworkState(responseError);
```
