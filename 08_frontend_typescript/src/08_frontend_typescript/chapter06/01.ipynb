## 6장 1강 : 제네릭 함수와 제네릭 인터페이스

### 1. 제네릭(Generic)의 개념과 필요성
다양한 타입을 하나의 틀로 다루면서도 타입 안전성을 보장하는 TypeScript의 타입 변수 시스템입니다.

#### 1.1 any 타입의 한계와 제네릭의 필요성
- any를 사용하면 컴파일러의 타입 검사가 비활성화되어 런타임 오류가 발생할 수 있습니다.
- 타입별로 개별 함수를 선언하면 동일한 로직이 반복되어 유지보수가 어려워집니다. (코드 중복 문제)
- 타입을 미리 고정하지 않고, 함수나 인터페이스를 호출하는 시점에 타입을 인자로 전달받아 결정합니다. (제네릭의 해결 방식)



any 타입과 제네릭의 비교

 any 타입을 사용한 함수 (타입 안전성 없음)

```typescript
function echoAny(value: any): any {
  // 입력받은 값을 그대로 반환하는 함수
  
  return value.toUpperCase();
}

```

제네릭을 사용한 함수 (타입 안전성 확보)

```typescript
function echoGeneric<T>(value: T): T {
  // <T>는 타입 매개변수로, 입력받은 value의 타입을 T로 지정하고 동일한 T 타입을 반환합니다.
  return value;
}

const num = echoGeneric<number>("100");
console.log("typeof num:", typeof num);

// T U V ... 알파벳 순서 혹은 매개변수 타입 value는 V, 반환값은 R 등 관례적 사용 명명이 있음
// <number> -> "100" -> Argument of type 'string' is not assignable to param of type 'number'
// 공통적인것 외에는 타입 좁히기가 필요함.\
```

```typescript
const num2 = echoGeneric<number>(100);
console.log("typeof num2:", typeof num2);
```

### 2. 제네릭 함수(Generic Function) 설계
함수를 호출할 때 타입을 인자로 전달하여 매개변수와 반환값의 타입을 동적으로 정하는 함수입니다.

#### 2.1 타입 매개변수 <T>와 타입 추론

- 타입 매개변수 식별자<br>
관례적으로 Type의 약자인 `<T>`를 사용하며, 여러 개가 필요한 경우 `<T, U, V>` 순으로 확장합니다.
- 타입 자동 추론<br>
함수 호출 시 <number>처럼 타입을 명시하지 않아도, 전달되는 인자 값을 보고 컴파일러가 타입을 자동으로 추론합니다

배열 요소를 처리하는 제네릭 함수 작성

```typescript
function getFirstElement<T, U>(arr: T[], info: U): T | undefined {
    if (arr.length === 0) {
        return undefined; //undefined 쓰지 않아도 반환값이 없으면 기본값이 이렇게 배정됨. 
    }

    console.log("Info:", info);

    return arr[0];
}

const numbers = [10,20,30];
// const firstNumber = getFirstElement<number, string> (numbers, "출력");
const firstNumber = getFirstElement<string> (numbers, "출력");
console.log(firstNumber)

// generic type사용은 interface로 사용해야함!
```

### 3. 제네릭 인터페이스(Generic Interface)와 응답 데이터 모델링

#### 3.1 공통 API 데이터 응답 구조 설계

```typescript
// 서버 응답을 위한 공통 제네릭 인터페이스 정의
interface ApiResponse<T> {
    success: boolean;   // 요청 성공 여부 (고정 타입)
    message: string;    // 응답 메시지 (고정 타입)
    data: T;            // 실제 데이터 (동적 타입 매개변수)
}

// 개별 데이터 구조 정의
interface UserProfile {
    id: number;     // 사용자 고유 번호
    name: string;   // 사용자 이름
    email: string;  // 사용자 이메일
}

interface ProductItem {
    code: string;       // 상품 코드
    price: number;      // 상품 가격
}

// UserProfile 타입을 가질 수 있는 응답 객체 생성
const userResponse : ApiResponse<UserProfile> = {
    success: true,
    message: "사용자 정보 조회 성공",
    data: {
        id: 1,
        name: "이름이",
        email: "user01@test.org"
    }
}
console.log("userResponse:", userResponse);

// ProductItem 타입을 가질 수 있는 응답 객체 생성
const productResponse: ApiResponse<ProductItem> = {
    success: true,
    message: "상품 정보 조회 성공",
    data: {
        code: "PROD-1000",
        price: 2000
    }
}
console.log("productResponse:", productResponse);
// 데이터 접근 및 타입 안전성 검증
```
