```python
from langchain.chat_models import init_chat_model
from langchain_core.prompts import PromptTemplate

model = init_chat_model(
    model="gpt-6-luna",
    model_provider="openai",
    reasoning_effort="low"
)

rewrite_prompt = PromptTemplate.from_template(
    # placeholder만 넣어주는 단순 형태
    """
    사용자의 질문을 검색 시스템이 문서를 탐색하기 유리하도록 핵심 명사 중심의 명확한 문장으로 다시 작성하세요.

    사용자의 질문: {question}
    변환된 질문: 
    """
)

rewrite_chain = rewrite_prompt | model
raw_query = "사과폰 최신형 베터리 얼마나 가?"
rewritten_query = rewrite_chain.invoke({"question": raw_query})

print(f"원래 질문: {raw_query}")
print(f"변환된 질문: {rewritten_query.content}")
```

```python
#HyDE

hyde_prompt = PromptTemplate.from_template(
    """
    다음 질문에 대한 객관적인 사실을 서술하는 한 단락의 학술적 답변을 작성하세요.

    질문: {question}
    가설 답변: 
    """
)

hyde_chain = hyde_prompt | model

hyde_query = "열이 펄펄 끓고 목이 따가울 때 먹는 약"
hyde_doc = hyde_chain.invoke({"question": hyde_query})

print(f"원래 질문: {hyde_query}")
print("[HYyDE 가설 문서]")
print(hyde_doc.content)
```
