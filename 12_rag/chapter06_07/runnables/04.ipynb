```python
# RunnablePassthrough(): 입력값을 그대로 전달

from langchain_core.runnables import RunnablePassthrough, chain

@chain
def func_a(data):
    data['text'] = f"**{data}**"

    return data

def func_b(data):
    data['text'] = f"##{data}##"

    return data
```

```python
chain = { "original_text": RunnablePassthrough()} | func_a | func_b

result = chain.invoke({"안녕하세요."})
print(result)
```
