```python
# 조건부 분기, RunnableBranch(함수, 함수의 결과가 참일때 실행되는 Rannable 객체, ..., Fallback Runnable 객체) 
# 조건이 간단하면 함수 별도 정의 안하고 람다 사용 가능
# 여러개 더 들어갈 수 있음 : ...
# 맨 마지막 모두 참이 아닐때 생성되는 것-> Fallback runnable 객체
# 연습은 람다로 진행함 

from langchain_core.runnables import RunnableLambda, RunnableBranch

branch = RunnableBranch(
    (lambda x: x == 'A', RunnableLambda(lambda x: f"A 입력")), # x 일단 넣음. 여기서는 필요 없을 수도!
    (lambda x: x == 'B', RunnableLambda(lambda x: f"B 입력")),
    RunnableLambda(lambda x: f"A, B 중에서 입력") # 마지막 함수를 넣어야 실행됨!
)

print(branch.invoke('A'))
print(branch.invoke('B'))
print(branch.invoke('C'))
```
