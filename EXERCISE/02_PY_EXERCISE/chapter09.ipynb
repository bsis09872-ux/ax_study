```python
# 2 통계 및 가공 함수를 활용한 성적 데이터 요약

korean_lang = [85, 62, 92, 78, 55]

korean_lang_avg = sum(korean_lang) / len(korean_lang)

print(round(korean_lang_avg, 2))

```

```python
over70 = list(filter(lambda score: score >= 70, korean_lang))
print(over70)
```

```python
# 3 내장 함수를 활용한 대기자 명부 및 매칭 시스템 구성

names = ["이민서", "박준형"]
emails = ["minseo@ssu.ac.kr", "junhyung@daum.net"]

for i, (name, email) in enumerate(zip(names, emails), start=1):
    print(f"{i}번 당첨자: {name} | 이메일: {email}")
    
```

```python
# 1 random 모듈을 활용한 무작위 데이터 선택
import random

menu = ['피자', '치킨', '햄버거', '파스타']
print(random.choice(menu))
```

```python
luck = random.randint(1,100)
print(f"오늘의 행운 점수: {luck}")
```

```python
# 3 json 모듈을 활용한 데이터 직렬화
import json

book_info = {
    "도서명": "파이썬기초", "저자":"팀스파르타", "가격": 30000
}

dicttjson = json.dumps(book_info, ensure_ascii=False)

print(dicttjson, type(dicttjson))
```
