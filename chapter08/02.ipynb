# 8장 2강: 상속과 캡슐화

1. 상속(inheritance)
- 클래스 기반의 상속 (상속에 대한 정의 -> 객체 생성(상속 관계 성립))

```
class 자식클래스(부모클래스):
    ...
```
- `self` 키워드
    - 자식 객체(현재 객체)의 참조 변수
- `super`키워드 (`self`키워드와 비슷함)
    - 부모 클래스를  가리키는 참조 변수
    - super(): 부모쪽 객체 생성
    
    - 부모클래스: 일반적인 정의, 상위 개념
    - 자식클래스: 구체적인 정의, 하위 개념


````python
class Animal:

    def __init__(self, kind):
        self.kind =  kind # 동물종류
        
    def move(self):
        print("동물이 움직인다.")

class Dog(Animal):
    def __init__(self):
        super().__init__("강아지") # 부모쪽 인스턴스 변수인 kind에 초기화

class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

````

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
````

2. 메서드 오버라이딩(overriding)
- 메서드 재정의
- 부모쪽 클래스와 동일한 이름으로 정의 + 기능을 수정 및 재정의
- 자식쪽 클래스에 정의된 동일한 이름의 함수가 호출 우선순위를 가진다.
- 자식쪽에서 부모쪽 동일한 이름의 메서드를 호출해야하는 경우는 self로 접근 x
    - super().부모쪽 메서드(...)

````python
class Animal:

    def __init__(self, kind):
        self.kind =  kind # 동물종류
        
    def move(self):
        print("동물이 움직인다.")

class Dog(Animal):
    def __init__(self):
        super().__init__("강아지") # 부모쪽 인스턴스 변수인 kind에 초기화

    def move(self):
        super().move() # 부모쪽 객체의 move() 호출
        print(self.kind + "걸음으로 움직인다.")

class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        super().move() # 부모쪽 객체의 move() 호출
        print(self.kind + "걸음으로 움직인다.")
````

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()

````

````python
class Human:
    def move(self):
        print("두 발로 직립 보행 한다.")
````

````python
human = Human()
human.move()
````

#### 다중상속 - 여러 부모클래스를 상속 관계로 정의 가능
- 다중 상속시 같은 메서드가 있을 때 호출 우선 순위(MRO)
- 자식클래스의 메서드 -> 상속 관계에서 왼쪽에 있는 클래스 -> 

````python
class ParentA:
    def a(self):
        print("ParentA - a()")

    def b(self):
        print("ParentA - b()")

class ParentB:
    def a(self):
        print("ParentB - a()")

    def c(self):
        print("ParentB - c()")

class Child(ParentB, ParentA):
    pass
````

````python
c = Child()
c.b()
c.c()

c.a()
````

````python
Child.__mro__ # method resolution order - 메서드 우선순위
````

4. 정보 은닉과 캡슐화(encapsulation)

````python
class Schedule:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day
        
    def __str__(self):
        return f"year = {self.year}, month = {self.month}, day = {self.day}"
````

````python
s1 = Schedule(2026,8,18)

print(s1)
````

````python
# 인스턴스 변수에 직접 대입이 가능하므로 값에 대한 통제는 불가!
# 통제 불가능하면 직접 접근을 막아야 한다. -> 인스턴스 변수를 외부에서 접근하지 못하게 막는다.

s1.year = 2026
s1.month = 2
s1.day = 31

print(s1)
````

5. 파이썬 프라이빗(private)변수와 이름 변형 규칙(name mangling)
- \_\_변수명: 클래스 내부에서만 접근 가능한 인스턴스 변수가 된다.

````python
class Schedule:
    def __init__(self,year,month,day):
        self.__year = year
        self.__month = month
        self.__day = day

    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
````

````python
s1 = Schedule(2026,8,18)
print(s1)
````

6. Getter와 Setter 메서드

````python
class Schedule:
    def __init__(self,year,month,day):
        self.__year = year
        self.__month = month
        self.__day = day

    # 통제 가능한 수단을 설정하기
    def get_year(self):
        return self.__year
    
    def get_month(self):
        return self.__month
    
    def get_day(self):
        return self.__day

    def set_year(self, year):
        self.__year = year

    def set_month(self, month):
        self.__month = month

    def set_day(self, day):
        if self.__month == 2 and day > 28: # 값에 대한 통제 가능
            self.day = 28 

        self.__day = day

    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
````

````python
s1 = Schedule(2026,8,18)
s1.set_day(31)
s1.set_month(2)

print(s1)
````
